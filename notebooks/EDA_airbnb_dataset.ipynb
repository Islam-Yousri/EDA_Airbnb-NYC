```python
#- Defining cleanNameFtr(x) Where x is value in name Feature 
def cleanNameFtr(val): 
    if '~' in str(val):
        val=val.replace('~','')
    if '!' in str(val):
        val=val.replace('!' , '')
    if '@' in str(val):
        val=val.replace('@' , '')
    if '#' in str(val):
        val=val.replace('#' , '')
    if '$' in str(val):
        val=val.replace('$' , '')
    if '%' in str(val):
        val=val.replace('%' , '')
    if '^' in str(val):
        val=val.replace('^' , '')
    if '&' in str(val):
        val=val.replace('&' , 'and')
    if '*' in str(val):
        val=val.replace('*' , '')
    if '(' in str(val):
        val=val.replace('(' , '')
    if ')' in str(val):
        val=val.replace(')' , '')
    if '-' in str(val):
        val=val.replace('-' , ' ')
    if '_' in str(val):
        val=val.replace('_' , ' ')
    if '+' in str(val):
        val=val.replace('+' , 'and')
    if '=' in str(val):
        val=val.replace('=' , 'equivalent')
    if '\\' in str(val):
        val= val.replace('\\' , ' ') 
    if '/' in str(val):
        val= val.replace('/' , 'and')
    if '?' in str(val):
        val= val.replace('?' , '')
    if ',' in str(val):
        val= val.replace(',' , ' ')
    if '.' in str(val):
        val= val.replace('.' , ' ')
    if 'new york' in str(val).lower():
        val = str(val).lower()
        val= val.replace('new york' , '')
    if '²' in str(val):
        val = val.replace('²','')

    return val
    
#- Convert 1,2,3,...n to 0ne,two,three,....,word
def convertNumToWord(val):
    for char in str(val):
        if char.isdigit():
            val = val.replace(char , num2words(int(char)))
    return val;

#- Executing Cleaning Process 
cleaned_renting_data['name'] = cleaned_renting_data['name'].apply(cleanNameFtr)
cleaned_renting_data['name'] = cleaned_renting_data['name'].apply(convertNumToWord)
```

```python
#- property_keywords in name Feature values 
def keywordCount(keywords_ls:list,feature_name:str):
    """
        - keywordCount Is A Function Is Used For Counting Number Of Listing Descriptions Have A Given Keyword
        - params:
          #- keywords_ls  : represents list of keywords 
          #- feature_name : represents the name of feature 
        - return 
           - A Dictionary Is Returned Such That Its keys Are keywords And its values Are count of keyword Appearance 
    """
    if not(str(renting_dataset[feature_name].dtype) == 'str') :
        print(f"The Values Of {feature_name} Feature Are Not String")
    else:
        dict_ = {}
        for p_k in keywords_ls:
            count=0
            dict_[p_k]=0 
            for name in renting_dataset[feature_name]:
                if p_k.lower() in str(name).lower():
                    count+=1
            dict_[p_k]+=count
        return dict_
```

```python
def checkKeywordExisting(desc:str,property_keywords:list):
    """
    - checkKeywordExisting Func Is Used For Extracting New Feature Holding Only Two Binary Values (1,0)
    - return value : 
        0 : It Is Main , The Description Of Listing (Accomodation) Hasn't Any Given Keyword 
    """
    for p_k in property_keywords:
        if str(p_k).lower() in str(desc).lower():
            return 1
    return 0 


cleaned_renting_data['has_prop_keyword']=renting_dataset['name'].apply(
    lambda desc: checkKeywordExisting(desc,property_keywords)
)
if 'has_prop_keyword' in cleaned_renting_data.columns.to_list():
    print("'has_prop_keyword' Is Found")
```

```python
#- defining cleanHostNameFtr(val) 
def cleanHostNameFtr(val):
    modified_str = str(val).lower()
    if '!' in str(modified_str): modified_str = modified_str.replace('!','') 
    if '@' in str(modified_str): modified_str = modified_str.replace('@','')
    if '&' in str(modified_str): modified_str = modified_str.replace('&','and') 
    if '(' in str(modified_str): modified_str = modified_str.replace('(',' and ')
    if ')' in str(modified_str): modified_str = modified_str.replace(')','')
    if '-' in str(modified_str): modified_str = modified_str.replace('-',' ')
    if '_' in str(modified_str): modified_str = modified_str.replace('_',' ')
    if '+' in str(modified_str): modified_str = modified_str.replace('+',' and ')
    if '/' in str(modified_str): modified_str = modified_str.replace('/',' and ')
    if ',' in str(modified_str): modified_str = modified_str.replace(',',' and ')
    if '.' in str(modified_str): modified_str = modified_str.replace('.','')

    #- Check If There is Digit In name And Remove It 
    chars = []
    for char in modified_str:
        if not char.isdigit():
            chars.append(char)
    modified_str = "".join(chars) 
    
    return modified_str
```

```python
#- Cleaning host_name Feature
cleaned_renting_data['host_name'] = cleaned_renting_data['host_name'].apply(cleanHostNameFtr)
```

```python
#- Remove All Neighbourhoods Associated With Places Where Listing Is listed 
def cleanBoroughs(val):
    
    boroughs = ["Manhattan" , "Brooklyn" , "Queens" , "Bronx","Staten Island"]
    for borough in boroughs:
        if borough in val:
            place = "".join(val.split(borough)).strip() 
            cleaned_place = [char for char in place  if not(char==',' or char=='-')]
            return cleaned_place 
    return val
cleaned_renting_data['neighbourhood'] =  cleaned_renting_data['neighbourhood'].apply(cleanBoroughs)
```

```python
#- Creating An Feature Representing accomodation_count_case 
   #- case: 1 --> single 
   #- case: 2=<accomodations<=25 --> small 
   #- case: 26=<accomodations<=50 --> medium 
   #- case: accomodations>=51 --> large  
def createAccomCountCase(num):
    if num == 1 :return "single"
    elif 2<=num<=25:return "small"
    elif 26<=num<=50:return "medium"
    elif num>=51:return "large"
        
cleaned_renting_data['accomodation_count_case']= renting_dataset['calculated_host_listings_count'].apply(createAccomCountCase)
if 'accomodation_count_case' in cleaned_renting_data.columns:
    print("'accomodation_count_case' is found")
```

##### <font color="1A05A2">***Data Engineering***</font>    

```python
#- Create A New Feature Representing Case Of Days Available For Renting 
 #- 0       --> un-avialable 
 #- 0-29    --> avialable_less_month 
 #- 30-179  --> avialable_greater_month 
 #- 180     --> avialable_half_year 
 #- 181-364 --> avialable_above_half-year 
 #  365     --> avialable_full_year
def avialabilityPeriod(val):
    if   val == 0 : return "un-available"
    elif 0<val<=29: return "avialable_less_than_month"
    elif 30<=val<=179: return "avialable_greater_than_month"
    elif val==180:return "avialable_half_year"
    elif 181<=val<=364:return "avialable_above_half-year"
    elif val<=365:return "avialable_full_year" 
cleaned_renting_data['availability_case' ] = renting_dataset['availability_365'].apply(avialabilityPeriod)
if 'availability_case' in cleaned_renting_data.columns:
    print("'availability_case' is found")
```
